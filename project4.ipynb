```python

import csv
import os
import smtplib
import ssl
from email.message import EmailMessage


# --------------------------------------------------
# STEP 1: Read students from CSV
# --------------------------------------------------

def read_students(filename):
    students = []

    with open(filename, "r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for student in reader:
            students.append(student)

    return students


# --------------------------------------------------
# STEP 2: Send one email
# --------------------------------------------------

def send_email(sender, password, recipient, subject, message):

    email = EmailMessage()

    email["From"] = sender
    email["To"] = recipient
    email["Subject"] = subject

    email.set_content(message)

    context = ssl.create_default_context()

    with smtplib.SMTP_SSL(
        "smtp.gmail.com",
        465,
        context=context
    ) as server:

        server.login(sender, password)
        server.send_message(email)


# --------------------------------------------------
# STEP 3: Main program
# --------------------------------------------------

def main():

    print("================================")
    print("      EMAIL AUTOMATION TOOL")
    print("================================")

    # Read CSV
    students = read_students("students.csv")

    print(f"\nFound {len(students)} students.")

    # Ask user for email details
    subject = input("\nEnter email subject: ")

    message = input(
        "Enter email message "
        "(use {name} for personalization): "
    )

    # Email account
    sender = input("Enter your Gmail address: ")

    # Get password from environment variable
    password = os.getenv("EMAIL_PASSWORD")

    if not password:
        print(
            "\nERROR: EMAIL_PASSWORD environment variable "
            "is not set."
        )
        return

    # Send emails
    for student in students:

        name = student["name"].strip()
        recipient = student["email"].strip()

        # Personalize message
        personalized_message = message.format(
            name=name
        )

        print(f"\nSending email to {name}...")

        try:

            send_email(
                sender,
                password,
                recipient,
                subject,
                personalized_message
            )

            print(
                f"Email successfully sent to "
                f"{recipient}"
            )

        except Exception as error:

            print(
                f"Could not send email to "
                f"{recipient}"
            )

            print("Error:", error)


# --------------------------------------------------
# STEP 4: Start program
# --------------------------------------------------

if __name__ == "__main__":
    main()

```
